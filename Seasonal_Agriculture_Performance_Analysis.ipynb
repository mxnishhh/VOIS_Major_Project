## 🧹 7. Data Cleaning & Preparation

The following checks are performed:

- Missing-value analysis
- Duplicate-record analysis
- Data-type verification
- Whitespace cleanup for text fields
- Duplicate removal
- Basic range validation for key numeric variables

We avoid blindly replacing missing values because inappropriate imputation can distort agricultural findings.

```python
# 🧹 Data Cleaning

# Standardize column names for easier Python use
df.columns = (
    df.columns.str.strip()
    .str.lower()
    .str.replace(r"[^a-z0-9]+", "_", regex=True)
    .str.strip("_")
)

# Remove unnecessary whitespace in text fields
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].apply(lambda x: x.strip() if isinstance(x, str) else x)

# Missing values
missing = pd.DataFrame({
    "Missing Values": df.isna().sum(),
    "Missing %": (df.isna().mean() * 100).round(2)
}).sort_values("Missing %", ascending=False)

print("🔎 Missing-Value Report")
display(missing)

# Duplicate check and removal
duplicates = df.duplicated().sum()
print(f"\n🔁 Duplicate rows found: {duplicates}")

df = df.drop_duplicates().reset_index(drop=True)

print(f"✅ Final cleaned dataset: {df.shape[0]:,} rows × {df.shape[1]} columns")
```

```python
# 🧪 Check for impossible / suspicious negative values

numeric_cols = df.select_dtypes(include=np.number).columns.tolist()

negative_counts = (df[numeric_cols] < 0).sum().sort_values(ascending=False)
negative_counts = negative_counts[negative_counts > 0]

if len(negative_counts):
    print("⚠️ Numeric columns containing negative values:")
    display(negative_counts.to_frame("Negative Values"))
else:
    print("✅ No negative numeric values detected.")

print("\n📌 Unique values by categorical column:")
categorical_cols = df.select_dtypes(include="object").columns.tolist()
for col in categorical_cols:
    print(f"{col}: {df[col].nunique()} unique values")
```

# 📈 PART I — STATISTICAL ANALYSIS

## One-Way ANOVA

### Hypotheses

**H₀:** Mean yield is equal across seasons.

**H₁:** At least one seasonal mean yield is different.

A **p-value < 0.05** is treated as evidence of a statistically significant difference at the 5% significance level.

> ⚠️ Statistical significance indicates an association/difference in the observed groups; it does not by itself prove causation.

```python
# 🧮 One-way ANOVA: Yield across seasons

groups = [
    group["yield_tonnes_ha"].dropna().values
    for _, group in df.groupby("season")
]

groups = [g for g in groups if len(g) >= 2]

f_stat, p_value = stats.f_oneway(*groups)

print(f"F-statistic : {f_stat:.4f}")
print(f"p-value     : {p_value:.6f}")

if p_value < 0.05:
    print("✅ Result: There is evidence of a statistically significant difference in yield across seasons.")
else:
    print("ℹ️ Result: No statistically significant seasonal yield difference was detected at the 5% level.")
```

# 🚨 PART J — UNUSUAL PATTERNS & OUTLIERS

## Outlier Detection

The IQR method is used:

**Lower bound = Q1 − 1.5 × IQR**

**Upper bound = Q3 + 1.5 × IQR**

Observations outside these limits are flagged as potential outliers.

```python
q1 = df["yield_tonnes_ha"].quantile(0.25)
q3 = df["yield_tonnes_ha"].quantile(0.75)
iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

yield_outliers = df[
    (df["yield_tonnes_ha"] < lower) |
    (df["yield_tonnes_ha"] > upper)
]

print(f"Q1: {q1:.3f}")
print(f"Q3: {q3:.3f}")
print(f"IQR: {iqr:.3f}")
print(f"Lower Bound: {lower:.3f}")
print(f"Upper Bound: {upper:.3f}")
print(f"🚨 Potential yield outliers: {len(yield_outliers):,}")

display(yield_outliers[
    ["farm_id", "state", "district", "crop", "season",
     "yield_tonnes_ha", "production_tonnes", "profit_inr"]
].head(20))
```

## 🌦️ Unusual Seasonal Patterns

A season is flagged when its mean yield is more than one overall standard deviation away from the overall dataset mean.

This is a **descriptive flag**, not proof that the season is abnormal.

```python
season_yield = df.groupby("season")["yield_tonnes_ha"].mean()

overall_mean = df["yield_tonnes_ha"].mean()
overall_std = df["yield_tonnes_ha"].std()

unusual_seasons = season_yield[
    (season_yield - overall_mean).abs() > overall_std
]

print(f"Overall mean yield: {overall_mean:.3f}")
print(f"Overall standard deviation: {overall_std:.3f}")

display(unusual_seasons.to_frame("Seasonal Mean Yield"))
```
